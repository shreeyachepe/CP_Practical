```python
# If elif else statement for marks of the student
# 95 and above grade A, 85-94 grade B, 75-84 grade C, 65-74 grade D, below 65 grade F
marks = int(input("Enter your marks: "))
if marks >= 95:
    print("Grade A")
elif marks >= 85:
    print("Grade B")
elif marks >= 75:
    print("Grade C")
elif marks >= 65:
    print("Grade D")
else:
    print("Grade F")
```
